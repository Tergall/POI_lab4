```python
import numpy as np
from keras import Sequential
from keras.layers import Dense
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
from sklearn.preprocessing import OneHotEncoder, LabelEncoder
```

```python
df = pd.read_csv('texture_features.csv')
df.head()
```

```python
X = df.drop(columns='category').values
y = df['category'].values

print('X:', X.shape)
print('y:', y.shape, y[:5])
```

```python
label_encoder = LabelEncoder()
y_int = label_encoder.fit_transform(y)

print('Kategorie:', label_encoder.classes_)
print('y_int:', y_int[:5])
```

```python
onehot_encoder = OneHotEncoder(sparse_output=False)
y_onehot = onehot_encoder.fit_transform(y_int.reshape(-1, 1))

print('y_onehot:', y_onehot.shape)
print(y_onehot[:5])
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y_onehot, test_size=0.3)

print('Treningowy:', X_train.shape, y_train.shape)
print('Testowy:   ', X_test.shape, y_test.shape)
```

```python
model = Sequential()
model.add(Dense(10, input_dim=72, activation='sigmoid'))
model.add(Dense(3, activation='softmax'))

model.summary()
```

```python
model.compile(loss='categorical_crossentropy', optimizer='sgd', metrics=['accuracy'])
```

```python
history = model.fit(X_train, y_train, epochs=100, batch_size=10, shuffle=True)
```

```python
y_pred = model.predict(X_test)
y_pred_int = np.argmax(y_pred, axis=1)
y_test_int = np.argmax(y_test, axis=1)
cm = confusion_matrix(y_test_int, y_pred_int)
print(cm)
```

```python
print(f'Dokładność: {np.trace(cm) / cm.sum():.2%}')
```
